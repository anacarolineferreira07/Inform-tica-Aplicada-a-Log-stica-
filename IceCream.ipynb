```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("sakshisatre/ice-cream-sales-dataset")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd

# Listar arquivos no diretório baixado
files = os.listdir(path)
print("Arquivos encontrados:", files)

# Procurar por arquivos CSV
csv_files = [f for f in files if f.endswith('.csv')]
if csv_files:
    csv_path = os.path.join(path, csv_files[0])
    df = pd.read_csv(csv_path)
    print(f"Dataset carregado com sucesso de: {csv_path}")
    display(df.head())
else:
    print("Nenhum arquivo CSV encontrado no diretório do dataset.")
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import kagglehub

# Obter o caminho dinâmico correto do dataset baixado
try:
    path = kagglehub.dataset_download("sakshisatre/ice-cream-sales-dataset")
    csv_path = os.path.join(path, 'Ice Cream.csv')
    df = pd.read_csv(csv_path)
except Exception as e:
    print(f"Erro ao carregar o arquivo: {e}")

# Preparar as variáveis e treinar o modelo de regressão
X = df[['Temperature']]
y = df['Revenue']
model = LinearRegression()
model.fit(X, y)

# Obter os coeficientes
coef_angular = model.coef_[0]
intercept = model.intercept_

# Calcular previsões e R²
y_pred = model.predict(X)
r2 = r2_score(y, y_pred)

# Criar o texto da equação (formato Excel)
texto_equacao = f"y = {coef_angular:.2f}x + {intercept:.2f}\nR² = {r2:.4f}"

# Configurar o estilo do gráfico
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Criar o gráfico de dispersão com a linha de regressão linear
sns.regplot(data=df, x='Temperature', y='Revenue', scatter_kws={'alpha':0.6}, line_kws={'color':'red'})

# Adicionar a equação e o R² no gráfico
plt.text(x=df['Temperature'].min() + 2, y=df['Revenue'].max() - 100, s=texto_equacao, fontsize=12, bbox=dict(facecolor='white', alpha=0.8, boxstyle='round,pad=0.5'), color='red', fontweight='bold')

# Adicionar títulos e rótulos aos eixos
plt.title('Relação entre Temperatura e Receita com Equação e R² (Estilo Excel)', fontsize=14)
plt.xlabel('Temperatura (°C)', fontsize=12)
plt.ylabel('Receita ($)', fontsize=12)

# Exibir o gráfico
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

# Preparar as variáveis x (independente) e y (dependente)
X = df[['Temperature']]
y = df['Revenue']

# Inicializar e treinar o modelo de regressão linear
model = LinearRegression()
model.fit(X, y)

# Obter os coeficientes
coeficiente_angular = model.coef_[0]
intercepto = model.intercept_

print(f"Coeficiente Angular (Slope/Inclinação): {coeficiente_angular:.4f}")
print(f"Coeficiente Linear (Intercepto/Ajuste Inicial): {intercepto:.4f}")
print(f"\nEquação da reta: Receita = {coeficiente_angular:.4f} * Temperatura + ({intercepto:.4f})")
```
