### Análise de Regressão Linear - Dados de Nupcialidade (SEADE)
Vamos carregar e visualizar os dados de nupcialidade para realizar uma regressão linear simples utilizando as bibliotecas `pandas`, `seaborn` e `matplotlib`.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Exemplo de URL de um dos arquivos CSV do repositório SEADE Nupcialidade.
# Caso queira usar outro arquivo específico do portal, basta substituir o link abaixo.
url_dados = "https://repositorio.seade.gov.br/dataset/c28be59b-134d-4952-b9cf-2a4f4d2f00d8/resource/034ee5c6-b328-444f-b673-8d6268894df4/download/seade_nupcialidade_casamentos.csv"

try:
    # Carregando os dados (ajustando o separador comum em arquivos brasileiros se necessário)
    df = pd.read_csv(url_dados, sep=';', encoding='utf-8')
except Exception as e:
    print(f"Não foi possível carregar diretamente via URL. Criando um dataframe de exemplo baseado na estrutura do SEADE. Erro: {e}")
    # DataFrame fictício de exemplo simulando a taxa de nupcialidade ao longo dos anos
    dados_exemplo = {
        'Ano': [2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022],
        'Casamentos': [65000, 66200, 67100, 68000, 69500, 68800, 67200, 66900, 66100, 65300, 52000, 58000, 61000]
    }
    df = pd.DataFrame(dados_exemplo)

display(df.head())
```

```python
# Configurando o estilo do gráfico
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Definindo as colunas para o gráfico. Ajuste os nomes de acordo com o CSV carregado.
# No nosso exemplo/fallback usaremos 'Ano' no eixo X e 'Casamentos' no eixo Y
x_col = 'Ano' if 'Ano' in df.columns else df.columns[0]
y_col = 'Casamentos' if 'Casamentos' in df.columns else df.columns[1]

# Gerando o gráfico com a linha de regressão linear
sns.regplot(x=x_col, y=y_col, data=df, ci=95,
            scatter_kws={"color": "blue", "alpha": 0.6, "s": 80},
            line_kws={"color": "red", "linewidth": 2})

plt.title('Regressão Linear: Tendência de Nupcialidade ao Longo do Tempo', fontsize=14, pad=15)
plt.xlabel(x_col, fontsize=12)
plt.ylabel(y_col, fontsize=12)

plt.tight_layout()
plt.show()
```

### Regressão Linear: Idade Média ao Casar (Homens vs. Mulheres)
Análise da relação e correlação linear entre a idade média dos homens ao se casar (Eixo X) e a idade média das mulheres ao se casar (Eixo Y).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Criando um conjunto de dados simulando a evolução da idade média ao casar (anos recentes)
dados_idade_casamento = {
    'Ano': [2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022],
    'Idade_Media_Homens': [29.5, 29.8, 30.1, 30.4, 30.7, 31.0, 31.3, 31.6, 31.9, 32.2, 32.5, 32.8, 33.1],
    'Idade_Media_Mulheres': [26.8, 27.1, 27.3, 27.7, 28.0, 28.3, 28.6, 28.9, 29.2, 29.5, 29.8, 30.1, 30.4]
}

df_idades = pd.DataFrame(dados_idade_casamento)
display(df_idades.head())
```

```python
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# Gerando o gráfico de regressão linear
sns.regplot(
    x='Idade_Media_Homens',
    y='Idade_Media_Mulheres',
    data=df_idades,
    ci=95,
    scatter_kws={"color": "teal", "alpha": 0.7, "s": 100, "label": "Anos Observados"},
    line_kws={"color": "darkorange", "linewidth": 2.5, "label": "Reta de Regressão"}
)

plt.title('Regressão Linear: Idade Média dos Homens vs. Mulheres ao Casar', fontsize=14, pad=15)
plt.xlabel('Idade Média dos Homens ao Casar (X)', fontsize=12)
plt.ylabel('Idade Média das Mulheres ao Casar (Y)', fontsize=12)
plt.legend(loc="upper left")

plt.tight_layout()
plt.show()
```

### Adicionando a Equação da Reta e $R^2$ ao Gráfico
Vamos utilizar a biblioteca `scipy.stats` para calcular os coeficientes de regressão linear (inclinação, intercepto, valor R e p-valor) para colocar essas métricas diretamente no gráfico.

```python
import numpy as np
from scipy import stats

# Calculando a regressão linear estatística
slope, intercept, r_value, p_value, std_err = stats.linregress(df_idades['Idade_Media_Homens'], df_idades['Idade_Media_Mulheres'])
r_squared = r_value ** 2

# Criando a string com a equação e o R²
equacao_txt = f"Y = {slope:.3f}X + ({intercept:.3f})\nR² = {r_squared:.4f}"
print("Equação calculada:", equacao_txt.replace('\n', ' | '))
```

```python
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# Gerando o gráfico principal
sns.regplot(
    x='Idade_Media_Homens',
    y='Idade_Media_Mulheres',
    data=df_idades,
    ci=95,
    scatter_kws={"color": "teal", "alpha": 0.7, "s": 100, "label": "Anos Observados"},
    line_kws={"color": "darkorange", "linewidth": 2.5, "label": "Reta de Regressão"}
)

# Adicionando o texto da equação dentro do gráfico de forma estilizada
plt.text(
    x=df_idades['Idade_Media_Homens'].min() + 0.2,
    y=df_idades['Idade_Media_Mulheres'].max() - 0.5,
    s=equacao_txt,
    fontsize=12,
    color='darkred',
    weight='bold',
    bbox=dict(facecolor='white', alpha=0.8, edgecolor='gray', boxstyle='round,pad=0.5')
)

plt.title('Regressão Linear com Equação da Reta e R² de Ajuste', fontsize=14, pad=15)
plt.xlabel('Idade Média dos Homens ao Casar (X)', fontsize=12)
plt.ylabel('Idade Média das Mulheres ao Casar (Y)', fontsize=12)
plt.legend(loc="lower right")

plt.tight_layout()
plt.show()
```
